# Image → GLB with TRELLIS (free Colab GPU)

1. **Runtime → Change runtime type → T4 GPU → Save**
2. **Runtime → Run all**
3. Step 3 asks you to pick your image. Leave it empty (Cancel) to use the image named in `IMAGE_IN_REPO`.
4. The GLB downloads by itself at the end. Install takes 15–25 min; the model itself 3–5 min.

```python
# @title 1. Get the project
REPO = "rajeev169/image-to-glb"  # @param {type:"string"}
IMAGE_IN_REPO = "rider_white_bg.png"  # @param {type:"string"}
import os, subprocess
if not os.path.exists("/content/image-to-glb"):
    subprocess.run(["git", "clone", "--depth", "1", f"https://github.com/{REPO}.git", "/content/image-to-glb"], check=True)
%cd /content/image-to-glb
!nvidia-smi --query-gpu=name,memory.total --format=csv
```

```python
# @title 2. Install TRELLIS (15-25 min, once per session)
!python gpu/trellis_gpu.py install
```

```python
# @title 3. Choose image (Cancel = use IMAGE_IN_REPO)
from google.colab import files
image_path = IMAGE_IN_REPO
try:
    up = files.upload()
    if up:
        image_path = next(iter(up))
except Exception as e:
    print("No upload:", e)
print("Using:", image_path)
```

```python
# @title 4. Generate GLB and download
SEED = 0  # @param {type:"integer"}
TEXTURE = 1024  # @param [512, 1024, 2048] {type:"raw"}
!python gpu/trellis_gpu.py generate "{image_path}" -o output/model.glb --seed {SEED} --texture {TEXTURE}
from google.colab import files
files.download("output/model.glb")
```
